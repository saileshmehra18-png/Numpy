## Advance operations with business models

```python
# Data structure : [restaurent_id: 2021,2022,2023,2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], #paradise biryani
    [2, 120000, 140000, 160000, 190000], #Beijing bites
    [3, 200000, 230000, 260000, 300000], #Pizza hub
    [4, 180000, 210000, 240000, 270000], #Burger point
    [5, 160000, 185000, 205000, 230000], #Chai point
])

print("=== Zomato Sales analysys ===")
print("\n Sales_data shape:",sales_data.shape)
print("\n Sample data for 1st 3 restaurents: \n",sales_data[:3])
print("\n All the rows without first column: \n",sales_data[:,1:])
```

```python
#total sales per year

yearly_total = np.sum(sales_data[:,1:],axis = 0)
total = np.sum(yearly_total)
print(total)

#min sales per restaurent

min_sales = np.min(sales_data[:,1:],axis = 1)
print(min_sales)

#maximum sales per year

max_sales = np.max(sales_data[:,1:],axis = 0)
print(max_sales)

#Average sales per restaurent

avg_sales = np.mean(sales_data[:,1:],axis = 1)
print(avg_sales)

#cumulative sales

cumsum = np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)
```

```python
cumsum = np.cumsum(sales_data[:,1:],axis = 1)
print(cumsum)
print(np.mean(cumsum,axis=0))

plt.figure(figsize=(8,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumulative sales accross all restaurent")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
