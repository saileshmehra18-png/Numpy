Numpy arrays and basics

```python
import numpy as np
```

### creating array from list

```python
arr_1d = np.array([1,2,3,4,5])
print("arr_1d:", arr_1d)

arr_2d = np.array([[1,2,3],[4,5,6]])
print("arr_2d:", arr_2d)
```

### list vs numpy arrays

```python
py_list = [1,2,3]
print("python list multiplication: ", py_list*2)

np_array = np.array(py_list)
print("numpy array multiplication: ", np_array*2)

import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n list operation time:", time.time()-start)


start = time.time()
py_list = np.arange(1000000)*2 #this function let's us creaet a array upto this point
print("\n list operation time:", time.time()-start)
```

### Creating arrays from scratch

```python
zeroes = np.zeros((3,4))
print("zeroes arrays: \n", zeroes)

ones = np.ones((3,4))
print("\n ones arrays: \n", ones)
```

```python
full = np.full((2,2),7)
print("full array: \n",full)

random = np.random.random((2,3))
print("random array: \n",random)

sequence = np.arange(0,10,2)
print("sequence array: \n",sequence)
```

### vector,matrices and tensor

```python
vector = np.array([1,2,3])
print("vector :\n", vector )

matrix = np.array([[1,2,3],[4,5,6]])
print("matrix : \n", matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("tensor: \n",tensor)
```

### Arrays properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])
print("shape: ",arr.shape)
print("dimension", arr.ndim)
print("size", arr.size)
print("Dtype:",arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12)
print("original array",arr)

reshaped = arr.reshape((3,4))
print("reshaped array: \n",reshaped)

# returns view instead of copy
raveled = reshaped.ravel()
print("revled: \n",raveled)

#transpose
transpose = reshaped.T
print("transposed array: \n",transpose)
```
